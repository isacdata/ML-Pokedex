# Separação de datasets
#### O objetivo desse notebook é separar o dataset principal em datasets de treino e teste

## Import de frameworks

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import f1_score
```

## Separação

```python
data_raw = pd.read_csv('datasets/spi_indicators.csv')
data_raw.head(2)
```

```python
data_raw.info()
```

```python
data_raw.describe(include='all')
```

```python
data_raw.income.value_counts()
```

Excluirei as colunas iso3c e country por serem muito específicas, podendo ser um ponto fácil pro modelo adivinhar qual é o income correto. Também excluirei o income classificado como "Not classified".

```python
df_filt = data_raw.drop(columns=['iso3c', 'country']).copy()
df_filt = df_filt[df_filt['income'] != 'Not classified']
df_filt.head(2)
```

```python
df_filt.describe(include='all')
```

```python
df_filt.year.unique()
```

```python
len(df_filt.year.unique())
```

A divisão mais correta, olhando para os dados seria separar entre 2004 e 2019 para termos 80% dos dados e 20% para teste. É importante ressaltar que provavelmente a pandemia pode afetar o modelo, todavia, como o propósito é estudar o optuna, isso será considerado para a modelagem.

```python
dataset_train = df_filt[df_filt.year < 2020].copy()
dataset_test = df_filt[df_filt.year >= 2020].copy()
```

```python
dataset_train.to_csv('datasets/dataset_train.csv', index=False)
dataset_test.to_csv('datasets/dataset_test.csv', index=False)
```
